```python
print("\n--- Program 2: Question Answering ---")

# Using DistilBERT for faster processing in Colab
qa_model = pipeline("question-answering", model="distilbert-base-cased-distilled-squad")

context = "Information Extraction (IE) is an essential component of NLP that converts unstructured text into structured data."
question = "What does Information Extraction do?"

result = qa_model(question=question, context=context)
print(f"Question: {question}")
print(f"Answer: {result['answer']}")
```

```python
def frequency_summarizer(text):
    stop_words = set(stopwords.words("english"))
    words = word_tokenize(text.lower())

    # Frequency Analysis[cite: 2]
    freq_table = {}
    for word in words:
        if word not in stop_words and word.isalnum():
            freq_table[word] = freq_table.get(word, 0) + 1

    sentences = sent_tokenize(text)
    sentence_scores = {}
    for sentence in sentences:
        for word, freq in freq_table.items():
            if word in sentence.lower():
                sentence_scores[sentence] = sentence_scores.get(sentence, 0) + freq

    # Selecting the top 2 sentences for the summary[cite: 2]
    summary_sentences = sorted(sentence_scores, key=sentence_scores.get, reverse=True)[:2]
    return " ".join(summary_sentences)

print("\n--- Program 3: Text Summarization ---")
text_input = "Much of the world's info is stored in unstructured text form. IE enables the conversion of this text into structured data. IE also automates repetitive tasks like tagging entities."
print("Summary:", frequency_summarizer(text_input))
```

```python
print("\n--- Program 4: Machine Translation ---")

# Explicitly loading model and tokenizer to avoid KeyError: 'translation'[cite: 2]
model_nm = "Helsinki-NLP/opus-mt-en-fr"
tokenizer = AutoTokenizer.from_pretrained(model_nm)
model = AutoModelForSeq2SeqLM.from_pretrained(model_nm)

english_text = "Machine translation automatically translates text from one natural language into another."

# Tokenization and Generation (The core of NMT processing)[cite: 2]
inputs = tokenizer(english_text, return_tensors="pt")
outputs = model.generate(**inputs)
translated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)

print(f"English: {english_text}")
print(f"French: {translated_text}")
```
